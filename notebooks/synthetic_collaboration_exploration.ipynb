# NeuroSync synthetic collaboration signal exploration

This notebook is an exploratory software walkthrough using seeded **synthetic EEG-like signals**, not human recordings. Its figures describe constructed signals and derived statistics. They do not validate a physiological model, identify emotion or intention, establish interpersonal neural synchrony, or predict collaboration quality.

The two simulated streams share generator-controlled oscillatory components under this example condition. Coherence is reported as a descriptive frequency-specific association. This notebook performs no surrogate inference, confidence intervals, or multiple-comparison correction.

```python
from neurosync.data.synthetic_eeg import ExperimentalCondition, generate_synthetic_eeg
from neurosync.preprocessing.raw import synthetic_participant_to_raw
from neurosync.features.spectral import extract_spectral_features
from neurosync.collaboration.synchrony import compute_synthetic_coherence
from neurosync.modeling.synthetic_baseline import build_synthetic_condition_dataset, evaluate_synthetic_baselines_cv
from neurosync.visualization import plot_coherence, plot_spectral_features, plot_synthetic_signals
```

```python
# Explicit seed and parameters make this synthetic fixture reproducible.
data = generate_synthetic_eeg(sampling_frequency_hz=256, duration_seconds=20, n_channels=4, seed=2026, condition=ExperimentalCondition.HIGH_SHARED_COMPONENT)
raw_a = synthetic_participant_to_raw(data, 'A')
raw_b = synthetic_participant_to_raw(data, 'B')
print(f'Source: {data.data_source}; condition: {data.condition.value}; seed: {data.random_seed}')
print(f'Sampling frequency: {data.sampling_frequency_hz:g} Hz; channels: {data.n_channels}; unit: V')
```

```python
# A short time segment; figure title and axis identify synthetic data and units.
plot_synthetic_signals(data, participant='A', channels=(0, 1), duration_seconds=5)
```

```python
# Spectral estimates describe this fixture and depend on analysis settings.
spectral_a = extract_spectral_features(raw_a)
plot_spectral_features(spectral_a)
```

```python
# Coherence is a statistical association, not proof of interpersonal synchrony.
coherence = compute_synthetic_coherence(raw_a, raw_b)
coherence.band_mean_coherence
plot_coherence(coherence)
```

## Synthetic generator-condition baseline

The following classifier target is the generator's HIGH/LOW shared-component setting. It is not an annotation of human collaboration. The paired conditions share a seed and are grouped together; the standardized model fits its scaler within each training fold. The metrics and Wilson interval describe this synthetic simulation workflow only.

```python
# A small reproducible grouped cross-validation run for notebook exploration.
condition_data = build_synthetic_condition_dataset(n_sessions=8, seed=7000, duration_seconds=6, n_channels=2)
baseline_cv = evaluate_synthetic_baselines_cv(condition_data, random_state=31, n_splits=4)
baseline_cv.aggregate_metrics
```

## Interpretation and limitations

The generator deliberately injects shared components; the plot visualizes known construction choices. This is not evidence of human neural synchrony or a causal interpersonal process. Coherence depends on sampling, window length, overlap, preprocessing, and finite sample count and can be upward biased. No null or surrogate control is included here. Do not interpret these synthetic statistics as labels for cognitive or collaboration states.
